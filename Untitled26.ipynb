```python
import matplotlib.pyplot as plt

import seaborn as sns
import pandas as pd
import numpy as np
```

```python
np.random.seed(0)
df = pd.DataFrame({
    'category': ['A' , 'B' , 'C' , 'D' , 'E'],
    'VALUE': [0, 70, 30, 90, 60],
    'SALES': np.random.randint(100,500, 5),
    'PROFIT': np.random.randint(10,500, 5)
})

print(df)
```

```python
#area plot
plt.figure()
plt.fill_between(df['category'], df['VALUE'], color='green', alpha=0.25)
plt.title("rivya")
plt.xlabel("category")
plt.ylabel("VALUE")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_style("whitegrid") # apply seaborn style
plt.figure(figsize=(7,4))
plt.fill_between(df['category'], df['SALES'],
                 color='skyblue', alpha=0.5, label='Sales')
plt.fill_between(df['category'], df['PROFIT'],
                 color='orange', alpha=0.5, label='Profit')
plt.title("rivya")
plt.xlabel("Category")
plt.ylabel("Values")
plt.legend()
plt.show()
```

```python
#pie chart
plt.figure()
plt.pie(df['VALUE'], labels=df['category'], autopct='%1.2f%%')
plt.title("rivya")
plt.show()
```

```python
from numpy._core.defchararray import center
#donut chart
plt.figure()
plt.pie(df['VALUE'], labels=df['category'], autopct='%1.1f%%')
center_circle = plt.Circle((0,0), 0.70, fc='white')
fig = plt.gcf()
fig.gca().add_artist(center_circle)
plt.title("rivya")
plt.show()
```

```python
#boxplot
plt.figure()
sns.boxplot(x=df['VALUE'])
plt.title("rivya")
plt.show()
```

```python
# Heatmap (correlation)
plt.figure()
corr = df[['SALES', 'PROFIT']].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap Rivya')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt #select only numeric coloums
corr = df[['VALUE', 'SALES', 'PROFIT']].corr()
plt.figure(figsize=(5,4))
sns.heatmap(corr, annot=True)
plt.title("Heatmap rivya")
plt.show()
```

```python
#Bubble Plot (3 Variable Chart)
import matplotlib.pyplot as plt
plt.scatter(df['SALES'], df['PROFIT'], s=df['VALUE']*10) # 's' parameter controls bubble size
plt.title("rivya")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(6,4))
sns.scatterplot(x='SALES', y='PROFIT', size='VALUE',
                hue='VALUE', data=df,
                sizes=(50, 300), palette='viridis')
plt.title("Bubble Plot using Seaborn")
plt.show()
```
